#1. What are data structures, and why are they important?

Data structures are ways of organizing and storing data so they can be used efficiently.
They help in fast access, searching, insertion, deletion, and management of data.

#2. Difference between mutable and immutable data types
Mutable	Immutable
Can be changed after creation	Cannot be changed once created
Examples: list, dict, set	Examples: int, float, string, tuple
#3. Differences between lists and tuples
List	Tuple
Mutable	Immutable
Slower	Faster
Uses []	Uses ()
Good for dynamic data	Good for fixed data
#4. How do dictionaries store data?

Dictionaries store data in key–value pairs using a hash table, allowing O(1) average lookup time.

#5. Why use a set instead of a list?

Removes duplicates automatically

Faster membership checking (in)

Unordered & unique elements

#6. What is a string, and how is it different from a list?

A string is a sequence of characters (immutable).
A list is a collection of elements (mutable).

#7. How do tuples ensure data integrity?

Tuples are immutable, so values cannot be changed accidentally.

#8. What is a hash table, and how is it related to dictionaries?

A hash table stores data using a hashing function.
Python dictionaries internally use a hash table.

#9. Can lists contain different data types?

Yes, lists can store multiple types (int, string, float, etc.).

#10. Why are strings immutable?

Because they are stored in a fixed memory location for speed, security, and memory optimization.

#11. Advantages of dictionaries over lists

Fast lookup using keys

Data stored as key-value pairs

Better for structured data

#12. When is a tuple preferable over a list?

When data should not change

For fixed configurations

As dictionary keys (lists cannot be keys)

#13. How do sets handle duplicates?

Sets remove duplicates automatically because they store only unique values.

#14. How does the in keyword work differently for lists and dictionaries?

In lists → checks elements

In dictionaries → checks keys only, not values

#15. Can you modify tuple elements?

No, tuples are immutable. Only lists can be modified.

#16. What is a nested dictionary? Example?

A dictionary inside another dictionary.

student = {
    "name": "Ajaz",
    "scores": {"math": 90, "science": 85}
}

#17. Time complexity of accessing elements in a dictionary

O(1) on average (constant time).

#18. When are lists preferred over dictionaries?

When order matters

When indexing is needed

For sequential data

#19. Why are dictionaries unordered?

Because they store items using hashing, not positions.
Data retrieval does not follow insertion order (before Python 3.7).

#20. Difference between list and dictionary in data retrieval

List → retrieve using index

Dictionary → retrieve using key
